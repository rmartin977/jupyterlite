For this lesson we will examine the frequency content known as the spectrum of a sound. Specifically we will look at a recording of the middle C note played on the piano. This note has a fundamental frequency of 263 Hz.  This sound also contains other frequencies or  over tones called harmonics which we will investigate.    

* Note that frequency which is cycles/sec is measured in  Hertz abbreviated Hz.

---
<span style="color:blue;font-size:28px;">1. Execute the cell below to load necessary libraries for this lesson.</span> 

```python
# Libraries

import numpy as np
import scipy

from scipy import signal as sg
import IPython
import matplotlib.pyplot as plt
%config InlineBackend.figure_format ='retina' 
import soundfile as sf
from numpy import pi,sin,cos
```

---
<span style="color:blue;font-size:28px;">2. Execute the cell below to read in the sound file.</span> 

Our sound is stored in the digital file called c_note.wav. Click here for detail on  [Wave files](https://en.wikipedia.org/wiki/WAV)



     It is not necessary to understand the code below at this time. We read in the sound and get:
        1. x : The list of sample values for the sound.
        2. fs: The sample rate.
     
     Press the arrow on the audio player to play the sound.

```python
x, fs = sf.read('c_note.wav')
# fs = 4000
# fs = 16000
# Play the audio back
IPython.display.Audio(x,rate=fs)
```

---
The sound was recorded by sampling it @ 8000 samples a second. So the digital file contains the sample values. In a code cell a line that starts with pound sign is a comment and will not be executed.  In the cell above uncomment each line ( remove the pound sign) to change the sampling
rate to 4000 and then 16000 and the run the cell again and listen to the distorted sound. For example playing @ 4000 Hz sample rate will slow down the playback and frequency will sound lower.


---
We can compute the lenght of the sound in seconds by multiplying the number of samples (given by the  Python command len(x)) by the time between samples
(given by 1/fs).  Execute cell below to determine the lenght of the recording. You should get about 2.5 seconds.

```python
len(x)* 1/fs
```

---
<span style="color:blue;font-size:28px;">3. Execute the cell below to view the sound in time domain.</span> 

     Notice the following about this plot:
     
        1. We only show about 120 ms of the recording
        2. The signal is not sinusoidal - a smooth sine but appears periodic.
        3. The fundamental frequency is computed by measuring period of signal (distance betweeen the two red vertical lines).See your plot.
        4. From the plot we measure the  period to be about T = 3.8 ms  thus giving f = 1/T = 263 Hz. As we suspected.
           
           

```python
plt.figure(figsize=(12,3))
y = x[600:]
t = np.arange(len(x))/fs * 1000
plt.plot(t[:500],y[:500],'-',linewidth=1)
plt.grid()
plt.title(" Middle C  - 263 Hz")
plt.xlabel('time(ms)');
plt.axvline(20,c='red',alpha=.5,ls=":")
plt.axvline(23.8,c='red',alpha=.5,ls=":");
```

---
<span style="color:blue;font-size:28px;">4. Execute the cell below to view the sound in frequency domain.</span> 

     Notice the following about this plot:
     
        1. The sound is complex - it contains multiple frequencies.
        2. The strongest energy is at the fundamental frequency, which is 263 Hz.
        3. There are harmonics present, that is frequency components at multiples of the fundamental.
        4. The amplitudes of harmonics diminish as frequency gets larger.
           

```python
plt.figure(figsize=(12,3))
X = np.fft.rfft(x)
X = X/max(X)
freqs = np.fft.rfftfreq(len(x),1/fs)
plt.plot(freqs[:5000],np.abs(X[:5000]))
plt.grid()
vlines = [263*k for k in {1,2,3,4,5,6,7}]

for v in vlines:
    plt.axvline(v,c='red',alpha=.5,ls=":")
plt.title("c_note");
plt.xlabel("frequency")
```

---
<span style="color:blue;font-size:28px;">5. Execute the cell below to filter out the harmonics from our cnote and 
play the filtered sound.</span> 

     Notice the following from your result :
     
        1. The fundamental frequency now dominates the spectrum. The higher harmonics have been strongly reduced by the filter.
        2. The harmonics are what give the sound its quality.
                  

```python
b,a = sg.butter(10, 400, fs=fs)  # design 10th order butterworth low pass filter with cutoff freq of 400 Hz
filtered_signal = sg.lfilter(b,a,x)  # Filter our the harmonics
IPython.display.Audio(filtered_signal,rate=fs)
```

---
<span style="color:blue;font-size:28px;">6. Execute the cell below to view the spectrum of filtered signal.</span> 

     Notice the following about this plot:
     
        1. The spectrum show just one frequency compononent. We have removed ("filtered out") the harmonics.
        

```python
# Look at  spectrum of filtered signal
plt.figure(figsize=(12,3))
XX = np.fft.rfft(filtered_signal)
freqs = np.fft.rfftfreq(len(filtered_signal),1/fs)
plt.plot(freqs,np.abs(XX))

plt.grid()
plt.title("Spectrum of filtered_signal");
```

---
<span style="color:blue;font-size:28px;">7. Execute the cell below to view the filtered signal waveform.</span> 

     Notice the following about this plot:
     
        1. The plot shows our filtered signal is now just a sinusoid with one frequency!
        

```python
plt.figure(figsize=(12,3))
y =filtered_signal[600:]
t = np.arange(len(x))/fs * 1000
plt.plot(t[:500],y[:500],'-',linewidth=1)
plt.grid()
plt.title(" Filtered signal")
plt.xlabel('time(ms)');
```

---
<span style="color:blue;font-size:28px;">7. Execute the cell below to compare the sound of the filtered signal to a pure 263 Hz tone.</span> 

    The code below generates a pure 263 Hz tone. We compare the three sounds.

    1. piano c-note
    2. filtered c-note 
    3. 263 Hz one

```python
t = np.arange(0,1,1/8000)
tone = np.cos(2*pi*(263*t))


display(f"c - note - piano ")
display(IPython.display.Audio(x,rate=8000))
display("filtered signal")
display(IPython.display.Audio(filtered_signal,rate=8000))
display(" 263 Hz tone")
display(IPython.display.Audio(tone,rate=8000))
```
