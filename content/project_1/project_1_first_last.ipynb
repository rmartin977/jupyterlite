```

How to find a rating.  To see what rating a given user gave a given movie just enter:  ratings(movie_id,user_id)

Execute the following cell to see what rating user #14 gave to movie id 0.  We should see 2.5 stars was given.

Suppose we want to know what movie got the largest overall score.  That is, we total the ratings for each movie and determine which one got the largest result.  To do this we simply sum all the rows and find the largest outcome. Note that we store all of the totals in array called "scores".

````python
scores = ratings.sum(axis=1) # scores is an array of scores for each row
scores.shape # the size of array is the number of movies  9742
````

````python
scores[:10] # this displays the first 10 scores, notice movie ID-1 got a score of 377.5.
````

````python
scores.max().item() # this tells us the largest value in the scores array is 1404. 
#            But what movie does this correspond to? We use the numpy argmax function.
````

## The max function returns the maximum value in an array.  The argmax function will return the index or position in the array where the maximum occurs. execute the cell below and note that the maximum in scores occurs at index value 277.

````python
np.argmax(scores) # movie ID-49 go the largest score
````

````python
scores[277] # Yep ID-49 got 1404 stars
````

````python
#and the winner is:
titles[277]
````
