In this part of the exercise we will implement a simple movie recommender system using
cosine similarity. How can we use linear algebra to determine how "similar" two movies are?

We represent each movie by a **feature vector** whose entries are the ratings given to that movie by the users in our data set. Thus, each user provides one feature, and two movies are represented using the same coordinate system. If two movies tend to receive similar ratings from the same users, their feature vectors will point in similar directions. We can therefore compare the movies by comapring the feature vectors! We compute the cosine of the angle between the vectors: a cosine similarity close to 1 (theta = 0)indicates very similar rating patterns, while a value closer to 0 (theta = 90 degrees) indicates that the movies are less similar.

Recall the cosine of the angle between two vectors is given by:

$$cos(\theta) =\frac{\overrightarrow{a}\cdot\overrightarrow{b}}{\overrightarrow{||a||}\;\overrightarrow{||b||}}$$

If two vectors are normalized (divide the vector by its magnitude) then the cosine similarity between the two vectors
is computed by simply computing the dot product of the vectors.  This dot product returns
a scalar that will give us a measure of how similar the two vectors are. We thus can determine
how similar two movies are by forming the dot product of the normed feature vectors for each 
movie. The feature vectors are the rows of the ratings_matrix.


Let us compare the movies Toy_Story and Full_Metal_Jacket.  These movies should *not* be very similar.


````python
get_Id("Full")
````

````python
# first normalize the vector for each movie.  Note Full Metal Jacket has movie id #187

toy_story_norm = ratings[0]/norm(ratings[0])
Full_Metal_Jacket_norm = ratings[187]/norm(ratings[187])
````

````python
# now compute the similarity between the two movies and display the result.

similarity_1 = np.dot(toy_story_norm,Full_Metal_Jacket_norm)

print(f'The similarity measure between Toy Story and Full Metal Jacket is {similarity_1}')
````

````python
#  Now we compare the movies Star Wars #49  and Return of the Jedi #180.  These movies should be similar.

Star_Wars_norm = ratings[49]/norm(ratings[49])
Return_of_the_Jedi_norm = ratings[180]/norm(ratings[180])

# now compute the similarity between the two movies and display the result.
similarity_2 = np.dot(Star_Wars_norm,Return_of_the_Jedi_norm)

print(f'\nThe similarity measure between Star Wars and Return of the Jedi is {similarity_2}')
````

Now we write a python function called top_five that will take as an input the id# 
for a movie that a user likes and return a list of the 5 movies we recommend to the user.
These will be the five movies most "similar" to the input.

For example when the function is executed with with Star Wars #49 as input the output
generated is below. Of course the first movie in the list should be the movie itself.

1 Star Wars (1977)

2 Return of the Jedi (1983)

3 Raiders of the Lost Ark (1981)

4 Empire Strikes Back, The (1980)

5 Toy Story (1995)

The cell below contains the function defintion for top_five.  You do not need to understand all the code at this time to complete this assignment. We will go over this code in class.

````python
def top_five(m,i):

    '''
    This program will determine the 5 rows in matrix m
    that are most similar to the target row i.

    Cosine similartiy is used in determining similarity of rows.
    
 '''
    # this line will normalize all the rows in the matrix

    m = m/norm(m,axis=1).reshape(-1,1) 
    
    # this line will compute the similarity between the target movie and all the other movies
    # the result is stored in the array similarities

    similarities = np.dot(m,m[i]) 
    
    # Next we sort the indices and print out the top five.
    
    
    l =np.argsort(similarities)[::-1][:5]
    for i,k in enumerate(l):
        print(i+1,titles[k])
````

## Moving Average Filter

Recall the transformation:

$$y[n] = \frac{x[n] + x[n-1]}{2} $$

defines a moving average filter of length 2. Its impulse response is

$$h[n] = \left\{\frac{1}{2}, \frac{1}{2}\right\}.$$

An input signal $x[n]$ is filtered by convolving it with this impulse response:

$$y[n] = x[n] * h[n].$$

A moving-average filter can be used to reduce hissing (high frequency) noise in an audio signal, as demonstrated below. Increasing the filter length produces more filtering and smoothing action. The example below uses a moving-average filter of length 7.

## Notch Filter

Recall from our lecture the transformation:

$$y[n] = x[n] - \sqrt{2}\,x[n-1] + x[n-2]$$

defines the filter/operator $T$ by

$$T(x[n] )= x[n] - \sqrt{2}\,x[n-1] + x[n-2].$$

This is a notch filter: a specific input frequency is rejected, or "notched out," from the input. Its impulse response is

$$h[n] = \left\{1,-\sqrt{2},1\right\}.$$

An input signal $x[n]$ is filtered by convolving it with this impulse response:

$$y[n] = x[n] * h[n].$$

To determine which digital frequency is notched out, we find a basis for the kernel of $T$. A sequence $x[n]$ is in the kernel when $T(x[n])=0$, so we solve the homogeneous difference equation

$$x[n] - \sqrt{2}\,x[n-1] + x[n-2] = 0.$$


<u>See Example 3 in Section 4.8 in your textbook.</u>

Use the trial solution $x[n]=r^n$. Substitution gives

$$r^n-\sqrt{2}\,r^{n-1}+r^{n-2}=0.$$

Factoring out $r^{n-2}$ produces the auxiliary equation

$$r^2-\sqrt{2}\,r+1=0.$$

By the quadratic formula, its roots are

$$r=\frac{\sqrt{2}\pm i\sqrt{2}}{2}=\frac{1\pm i}{\sqrt{2}}.$$

Each root has magnitude $1$ and angle $\pm\pi/4$, so in polar/exponential form

$$r_{\pm}=e^{\pm i\pi/4}.$$

Therefore, over the complex numbers, the kernel is

$$\ker(T)=\operatorname{span}\left\{e^{i\pi n/4},e^{-i\pi n/4}\right\}.$$

Using Euler's formula, an equivalent real sinusoidal basis is

$$\ker(T)=\operatorname{span}\left\{\cos\left(\frac{\pi n}{4}\right),\sin\left(\frac{\pi n}{4}\right)\right\}.$$

Thus the notch, or rejected digital angular frequency, is

$$\boxed{\omega_0=\frac{\pi}{4}\text{ radians/sample}}.$$

For an audio sampling rate of $f_s=8000$ samples/second, convert digital angular frequency to analog frequency in hertz using

$$f_0=\left(\frac{\omega_0}{2\pi}\right)f_s=\left(\frac{\pi/4}{2\pi}\right)(8000)=1000\text{ Hz}.$$

See class notes for a derivation of this.

Therefore, the filter will be used below to remove a $1000$ Hz tone from an audio signal sampled at $8000$ samples/second.
