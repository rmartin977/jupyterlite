# Project 3. Digital Images and Matrices

A digital image is represented by a rectangular array of numbers called **pixels** (picture elements). In a grayscale image, each pixel has a numerical value representing its brightness or intensity. Typically, these values range from 0 (black) to 255 (white), with intermediate values representing shades of gray. The pixel values are stored in a matrix, where each entry corresponds to a particular location in the image. The **resolution** of an image is determined by the number of rows and columns in this matrix. For example, an image with a resolution of 640 × 480 pixels contains 640 columns and 480 rows, for a total of 307,200 pixels. A color image is represented using three intensity values for each pixel, corresponding to the red, green, and blue (RGB) color channels. Thus, a color image can be represented by three matrices, one for each color channel, or equivalently by a three-dimensional array.

# Part 2.  Low rank approximation of matrices/images.



The Real Symmetric Case

Recall the following facts about a real, symmetric n × n matrix A (where \(A^T = A\)):

• Real eigenvalues: All eigenvalues of A are real numbers.

• Orthogonal eigenvectors: Eigenvectors corresponding to distinct eigenvalues are mutually orthogonal.

• Orthonormal diagonalization: A can be factored as:

$$A=QDQ^{T}$$

The columns of $Q$ are orthonormal eigenvectors of $A$:

$$Q=\begin{bmatrix}\mathbf{q}_1 & \mathbf{q}_2 & \cdots & \mathbf{q}_n\end{bmatrix}.$$

The matrix $D$ is diagonal, with the corresponding eigenvalues of $A$ on its diagonal:

$$D=\begin{bmatrix}
\lambda_1 & 0 & \cdots & 0 \\
0 & \lambda_2 & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & \lambda_n
\end{bmatrix}.$$


The exact spectral decomposition of $A$ is

$$A=\sum_{i=1}^{n}\lambda_i\mathbf{q}_i\mathbf{q}_i^{T}.$$

Thus, $A$ is written as a sum of rank-one matrices. A rank-$k$ approximation of $A$ is obtained by retaining only $k$ of these terms:

$$A_k=\sum_{i=1}^{k}\lambda_i\mathbf{q}_i\mathbf{q}_i^{T}.$$

For the rank-$k$ approximation, the eigenvalues and their corresponding eigenvectors must first be ordered so that

$$|\lambda_1|\geq |\lambda_2|\geq\cdots\geq |\lambda_n|.$$

In other words, we retain the $k$ eigenvalues having the largest absolute values, together with their corresponding eigenvectors.

----

### Execute the cell below to load and display the image we call squares.

```python
squares = np.load("squares.npy")
plt.imshow(squares,cmap='gray')
plt.axis("off");
```

### Execute the cell below to compute the eigenvalues and eigenvectors for `squares` and sort them.

---

```python
w, Q = np.linalg.eigh(squares)
# Sort by descending absolute eigenvalue
idx = np.argsort(np.abs(w))[::-1]
w = w[idx]
Q = Q[:, idx]
```

###  Execute the cell below to compute and display a rank 1 approximation for the image.

---

```python
k=1
low_rank = np.sum([ w[i]*np.outer(Q[:,i],Q[:,i]) for i in range(k)],axis=0)

plt.imshow(low_rank,cmap='gray',vmin=0,vmax=255)
```
